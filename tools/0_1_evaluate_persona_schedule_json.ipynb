```python
import pandas as pd
from warnings import filterwarnings
filterwarnings("ignore")

# pd.set_option('display.max_colwidth', None)
pd.set_option('display.max_rows', None)
```

## 1. Load schedule JSON

- This JSON file is generated by 

Take whatever is inside each cell of df['persona'] and expand it into separate columns.

```python
seed_persona = pd.read_json('../app/src/smallville_personas.json')
gemini_json = pd.read_json('../app/src/driftville_personas_model_comparisons/driftville_personas_v8_gemini-3-flash-preview:cloud.json')
gemini_json

json_persona = pd.read_json('../app/src/driftville_personas.json')
json_persona

json_persona == gemini_json
```

Above, we just confimred that the final `driftville_personas.json` was generated by **Gemini 3 Flash Preview** model.

```python
# clean the raw persona text
seed_persona['raw_persona'] = seed_persona['raw_persona'].apply(lambda x: x.replace('\n', ' '))
seed_persona
```

```python
# Merge seed persona and generated json persona to compare 
# seed_person + JSON (persona)
merged = pd.merge(seed_persona, json_persona['persona'].apply(pd.Series), left_on='name', right_on='name', how='outer')
# seed_person + JSON (persona) + JSON (schedule)
merged = pd.concat([merged, json_persona.explode('schedule').apply(pd.Series)['schedule'].apply(pd.Series)], axis=1)
print(merged.shape)
merged.tail()
```

## Export merged to CSV

- smallville persona seed(JSON) + driftville persona/scheudle (JSON)

```python
merged.to_csv('../app/src/merged_driftville_persona.csv', index=False)
```
